```python
import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy.engine import URL
from getpass import getpass

# Read the CSV
df = pd.read_csv("Zomato.csv", encoding="latin-1")
df.columns = df.columns.str.strip().str.lower().str.replace(r"\s+", "_", regex=True)

# Connect to your existing database
password = getpass("Enter your MySQL root password: ")
connection_url = URL.create(
    "mysql+pymysql",
    username="root",
    password=password,
    host="localhost",
    database="zomato_analysis"
)
engine = create_engine(connection_url)

# Create a separate table so the old table stays untouched
df.to_sql(
    "restaurants_full",
    con=engine,
    if_exists="fail",
    index=False,
    chunksize=1000
)

print("Import complete:", len(df), "rows")
```

```python
import matplotlib.pyplot as plt
print("Matplotlib is ready")
```

```python
print("Rows and columns:", df.shape)

print("Duplicate rows:", df.duplicated().sum())

print("\nMissing values by column:")
print(df.isnull().sum().sort_values(ascending=False))

print("\nColumn names:")
print(df.columns.tolist())
```

```python
df_clean = df.copy()

df_clean["cuisines"] = df_clean["cuisines"].fillna("Not specified")

print("Missing cuisine values after cleaning:", df_clean["cuisines"].isnull().sum())
print("Cleaned dataset shape:", df_clean.shape)
```

```python
top_cities = df_clean["city"].value_counts().head(10)

top_cities.plot(kind="bar", figsize=(10, 5))
plt.title("Top 10 Cities by Restaurant Count")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("top_cities.png")
plt.show()
```

```python
df_clean.to_csv("Zomato_cleaned.csv", index=False)
```

```python
city_ratings = (
    df_clean[df_clean["aggregate_rating"] > 0]
    .groupby("city")["aggregate_rating"]
    .mean()
    .sort_values(ascending=False)
)

top_cities_by_count = df_clean["city"].value_counts().head(10).index
city_ratings = city_ratings.loc[city_ratings.index.isin(top_cities_by_count)]

city_ratings.plot(kind="bar", figsize=(10, 5))
plt.title("Average Restaurant Rating in Top 10 Cities by Count")
plt.xlabel("City")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.ylim(0, 5)
plt.tight_layout()
plt.savefig("ratings_by_city.png")
plt.show()
```

```python
delivery_counts = df_clean["has_online_delivery"].value_counts()

delivery_counts.plot(kind="bar", figsize=(7, 4))
plt.title("Restaurants by Online Delivery Availability")
plt.xlabel("Online Delivery")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("online_delivery_counts.png")
plt.show()
```

```python
price_counts = df_clean["price_range"].value_counts().sort_index()

price_counts.plot(kind="bar", figsize=(7, 4))
plt.title("Restaurant Count by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("restaurant_count_by_price_range.png")
plt.show()
```

```python
rated = df_clean[df_clean["aggregate_rating"] > 0]

rated["aggregate_rating"].plot(
    kind="hist",
    bins=20,
    figsize=(9, 5),
    edgecolor="black"
)

plt.title("Distribution of Restaurant Ratings")
plt.xlabel("Aggregate Rating")
plt.ylabel("Number of Restaurants")
plt.tight_layout()
plt.savefig("rating_distribution.png")
plt.show()
```

```python
delivery_by_price = (
    df_clean.assign(
        delivery_yes=(df_clean["has_online_delivery"] == "Yes")
    )
    .groupby("price_range")["delivery_yes"]
    .mean()
    .mul(100)
)

delivery_by_price.plot(kind="bar", figsize=(8, 5))

plt.title("Online Delivery Availability by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Restaurants Offering Delivery (%)")
plt.xticks(rotation=0)
plt.ylim(0, 100)
plt.tight_layout()
plt.savefig("delivery_percentage_by_price_range.png")
plt.show()
```

```python
rating_by_price = (
    df_clean[df_clean["aggregate_rating"] > 0]
    .groupby("price_range")["aggregate_rating"]
    .mean()
)

rating_by_price.plot(kind="bar", figsize=(8, 5))

plt.title("Average Restaurant Rating by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Average Rating")
plt.xticks(rotation=0)
plt.ylim(0, 5)
plt.tight_layout()
plt.savefig("average_rating_by_price_range.png")
plt.show()
```

```python
top_cuisines = df_clean["cuisines"].value_counts().head(10)

top_cuisines.plot(kind="bar", figsize=(11, 5))

plt.title("Top 10 Most Common Cuisine Entries")
plt.xlabel("Cuisine")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("top_10_cuisines.png")
plt.show()
```

```python
rating_by_delivery = (
    df_clean[df_clean["aggregate_rating"] > 0]
    .groupby("has_online_delivery")["aggregate_rating"]
    .mean()
)

rating_by_delivery.plot(kind="bar", figsize=(7, 4))

plt.title("Average Restaurant Rating by Online Delivery")
plt.xlabel("Online Delivery")
plt.ylabel("Average Rating")
plt.xticks(rotation=0)
plt.ylim(0, 5)
plt.tight_layout()
plt.savefig("average_rating_by_online_delivery.png")
plt.show()
```

```python
top_voted = df_clean.nlargest(10, "votes").copy()
top_voted["label"] = top_voted["restaurant_name"] + " (" + top_voted["city"] + ")"

top_voted = top_voted.sort_values("votes")

top_voted.plot(
    x="label",
    y="votes",
    kind="barh",
    figsize=(10, 6),
    legend=False
)

plt.title("Top 10 Restaurant Records by Votes")
plt.xlabel("Number of Votes")
plt.ylabel("Restaurant")
plt.tight_layout()
plt.savefig("top_10_restaurants_by_votes.png")
plt.show()
```
